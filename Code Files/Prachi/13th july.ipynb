```python
def print_info(**kwargs):
    dict = {}
    for key, value in kwargs.items():
        dict[key] = value
    return dict

print_info(a=1, b=2, c=3)
```

```python
def out_func():
    x="outer"
    def inner_func():
        nonlocal x
        x = "inner"
        print(x)
    inner_func()
    print(x)
out_func()
```

```python
def triangle_print(n):
    for i in range(1,n+1):
        print("*" * i)
```

```python
triangle_print(9)
```

```python
triangle_print(40)
```

```python
def print_right_triangle(n):
    for i in range(1,n+1):
        print(" " * (n-i) + "*" * i)
```

```python
print_right_triangle(6)
```
