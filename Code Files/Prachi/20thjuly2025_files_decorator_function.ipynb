What is a Decorator in Python?

A decorator is a function that takes another function as input, adds some functionality to it without modifying its structure, and returns a new function.

```python
@my_decorators
def greet():
    print("Hello")
```

```python
greet()
```

```python
def my_decorators(func):
    def wrapper():
        print("Before thye function")
        func()
        print("After the function")
    return wrapper
```

```python
@my_decorators
def add(a,b):
    return a+b
```

```python
def invite_friends(party_func):
    def wrapper(*args, **kwargs):
        print("Group message bhej diya: Bhai ka breakup party hai!")
        party_func(*args, **kwargs)
        print("Sab friends milke gale lage!")
    return wrapper

def decorate_venue(party_func):
    def wrapper(*args, **kwargs):
        print("Balloons lag gaye!")
        print("Lights decorate kar diye!")
        party_func(*args, **kwargs)
        print("Saaf safai ho gayi venue ki.")
    return wrapper

@invite_friends
@decorate_venue
def bhai_ka_party(reason):
    print(f"Bhai bolta hai: Aaj party hogi because {reason}")

bhai_ka_party("kaddu katega sab me batega")
```
